```python
import tensorflow as tf
from pathlib import Path
import json

json_path = Path.cwd().parent / "constructed_models" / "xdata_models" / "info.json"

with open(json_path, "r") as f:
    json_data = json.load(f)

model_path = Path.cwd().parent / "constructed_models" / "xdata_models"

for dir in model_path.iterdir():

    if dir.name == "info.json":
        continue

    metadata = json_data[dir.name.split('.')[0]]
    
    model = tf.keras.models.load_model(model_path / dir.name, safe_mode=False
    )
    print(f"Name: {dir.name}"); print()
    for k,v in metadata.items():
        print(k)
        print(v)
        print()
    model.summary()
    print()
```
